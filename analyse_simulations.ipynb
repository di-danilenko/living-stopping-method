```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('simulated_screening/Wassenaar_2017-0.9-results-1.csv')
df.head()
```

```python
fig, ax = plt.subplots()

ax.plot(df["sim_order"], df["incl"].cumsum()/df["incl"].sum())

ax.plot(df["sim_order"], df["buscar_p_r080"])
```
