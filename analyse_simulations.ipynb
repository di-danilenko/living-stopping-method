```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('simulated_screening/Bos_2018-0.99-results-1.csv')
df.head()
```

```python
fig, ax = plt.subplots()

ax.plot(df["sim_order"], df["incl"].cumsum()/df["incl"].sum())

ax.plot(df["sim_order"], df["buscar_p_r099"])
```

```python
df = pd.read_csv('simulated_screening_reordering/Bos_2018-0.99-results-1.csv')
fig, ax = plt.subplots()

ax.plot(df["sim_order"], df["incl"].cumsum()/df["incl"].sum())


ax.plot(df["sim_order"], df["buscar_p_r099"])
```

```python
df_reordered = df.sort_values('sim_reordered').reset_index(drop=True)

fig, ax = plt.subplots()

ax.plot(df_reordered["sim_reordered"], df_reordered["incl"].cumsum()/df["incl"].sum())

ax.plot(df["sim_order"], df["buscar_p_r099"])
```
